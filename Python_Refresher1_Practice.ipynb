**In Colab, and in many interactive Python environments like Jupyter notebooks, if a cell contains multiple expressions that would produce output, only the result of the last expression or statement is automatically displayed as the cell's output. Earlier expressions or statements that would normally produce output (like type(a) or simply a on its own line) are evaluated but their results are not implicitly rendered unless you explicitly use a print() function for them. The print() function, however, immediately outputs its argument to standard output, making it visible regardless of its position in the cell.**

### Sorting Lists

There are two primary ways to sort lists in Python in ascending order:

1.  **`list.sort()` method**: This method sorts the list in-place, meaning it modifies the original list and returns `None`.
2.  **`sorted()` function**: This built-in function returns a *new* sorted list from the items in an iterable, leaving the original list unchanged.

```python
# Using the list.sort() method (sorts in-place)
numbers = [4, 2, 6, 1, 9, 3, 5]
# below 3 print statements are giving same output but print expression is different
print(f"Original list: {numbers}")
print('Original list:',numbers)
print('Original List:', format(numbers))

#using ''' or """ can be considered as multi line comment in Python
'''This uses an f-string (formatted string literal).
The f before the opening quote indicates that it's an f-string. Inside the string
'''


numbers.sort() # Modifies the list in-place
print(f"Sorted list (using .sort()): {numbers}")

# Using the sorted() function (returns a new list)
other_numbers = [8, 5, 1, 7, 2, 0, 9]
print(f"\nOriginal list: {other_numbers}")

sorted_other_numbers = sorted(other_numbers) # Returns a new sorted list
print(f"Sorted list (using sorted()): {sorted_other_numbers}")
print(f"Original list remains unchanged: {other_numbers}")
```
